## Task 1: Check Divisibility
Ask the user to enter a number. If the number is divisible by both 3 and 5, print `FizzBuzz`.

```python
num=int(input('please enter a number: '))
if num%3==0 and num%5==0:
  print('FizzBuzz')
```

## Task 2: Even, Odd, or Zero
Ask the user to enter a number. Print:
- `Even` if the number is even and not zero
- `Odd` if it's odd
- `Zero` if the number is 0

```python
num1=int(input('please enter a number: '))
if num1%2==0 and num1>0:
  print('Even')
elif num1%2!=0:
  print('Odd')
else:
  print('Zero')
```

## Task 4: Compare Two Numbers
Ask the user to enter two numbers and print:
- `Equal` if they are the same
- `First is greater` if the first is greater
- `Second is greater` otherwise

```python
number1=int(input('please enter first number: '))
number2=int(input('please enter second number: '))
if number1==number2:
  print('Equal')
elif number1>number2:
  print('First is greater')
elif number1<number2:
  print('Second is greater')
```

## Task 5: Grade with Bonus
Ask for a score and whether the student has a bonus (`yes` or `no`).
If `yes`, add 5 points. Then display the grade:
- 90+: Excellent
- 75–89: Good
- 60–74: Pass
- Below 60: Fail

```python
score=int(input('please enter your score: '))
bonus=input('do you have a bonus? ')
if bonus=='yes':
  score+=5
  if score>=90:
    print('Excellent')
  elif score >= 75 and score <=89:
    print('Good')
  elif score >= 60 and score <=74:
    print('Pass')
  else:
    print('Fail')
```

## Task 7: Temperature Advice
Ask for the temperature in Celsius and print:
- `Very hot` if > 40
- `Cold` if < 10
- `Perfect weather` if between 20 and 30
- `Normal` otherwise

```python
temp=int(input('please enter temperature in Celsius: '))
if temp>40:
  print('Very hot')
elif 20 <= temp <= 30:
  print('Perfect weather')
elif temp<10:
  print('Cold')
else:
  print('Normal')
```

## Task 9: Arithmetic Checker
Ask for two numbers and an operator (`+`, `-`, `*`, `/`). Perform the operation using `if-elif` conditions.

```python
numm1=int(input('enter first number: '))
numm2=int(input('enter second number: '))
op=input('enter an operator (+,-,*,/): ')
if op=='+':
  result=numm1+numm2
  print(result)
elif op=="-":
  result=numm1-numm2
  print(result)
elif op=='*':
  result=numm1*numm2
  print(result)
elif op=='/':
  result=numm1/numm2
  print(result)
else:
  print('try again')
```

## Task 10: Ternary Challenge
