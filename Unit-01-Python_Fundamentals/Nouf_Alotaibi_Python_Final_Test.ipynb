### Q14 — Error Handling with try/except

Write a function `safe_divide(a, b)` that:

### Q17 — OOP: Class with Encapsulation

Create a class `BankAccount` with:
- **Attributes:** `owner` (string), `__balance` (private, default = 0)
- **Methods:**
  - `deposit(amount)` — adds the amount to balance (only if amount > 0)
  - `withdraw(amount)` — subtracts the amount from balance (only if amount > 0 and sufficient balance), otherwise print `"Insufficient balance."`
  - `get_balance()` — returns the current balance

**Test:**
1. Create an account for `"Ali"` with initial balance `1000`.
2. Deposit `500`.
3. Withdraw `200`.
4. Try to withdraw `2000` (should print insufficient balance).
5. Print the final balance.

````python
class BankAccount:
  def __init__(self, owner, balance=0):
    self.owner = owner
    self.__balance = balance

  def deposit(self, amount):
    if amount > 0:
      self.__balance += amount
    else:
      print("can't deposit 0.")

  def withdraw(self, amount):
    if amount > 0:
      if self.__balance >= amount:
        self.__balance -= amount
      else:
        print('Insufficient balance.')
    else:
      print('Amount must be positive.')

  def get_balance(self):
    return self.__balance

account1 = BankAccount('Ali', 1000)

account1.deposit(500)
account1.withdraw(2000)
print(account1.get_balance())
````

### Q18 — Inheritance and Polymorphism

Create the following class hierarchy:

1. A **base class** `Shape` with:
   - A method `area()` that returns `0`.
   - A method `describe()` that returns `"I am a shape."`

2. A class `Rectangle` that **inherits** from `Shape`:
   - Has `width` and `height` attributes.
   - Overrides `area()` to return `width * height`.

3. A class `Circle` that **inherits** from `Shape`:
   - Has a `radius` attribute.
   - Overrides `area()` to return `3.14 * radius ** 2`.

**Test polymorphism:**
- Create a list containing a `Rectangle(5, 4)` and a `Circle(7)`.
- Loop through the list and print each shape's `area()`.

````python
class Shape:
  def area(self):
    return 0

  def describe(self):
    return 'I am a shape'

class Rectangle(Shape):
  def __init__(self,width,height):
    self.width=width
    self.height=height

  def area(self):
    return self.width*self.height

class Circle(Shape):
  def __init__(self,radius):
    self.radius=radius

  def area(self):
    return 3.14*self.radius**2

shapes=[Rectangle(5,4),Circle(7)]
for shape in shapes:
  print(shape.area())
````

### B1 — Abstract Classes and Polymorphism

Using the `abc` module:

1. Create an **abstract class** `Vehicle` with:
   - An abstract method `start()`
   - An abstract method `stop()`

2. Create a class `ElectricCar` that inherits from `Vehicle`:
   - Has an attribute `brand`.
   - Implements `start()` → prints `"<brand> electric engine starting silently..."`
   - Implements `stop()` → prints `"<brand> has stopped. Battery saved."`

3. Create a class `GasCar` that inherits from `Vehicle`:
   - Has an attribute `brand`.
   - Implements `start()` → prints `"<brand> gas engine roaring to life!"`
   - Implements `stop()` → prints `"<brand> engine turned off."`

**Test:**
- Create a list with an `ElectricCar("Tesla")` and a `GasCar("Toyota")`.
- Loop through the list and call `start()` and `stop()` on each — demonstrating **polymorphism**.

````python
from abc import ABC, abstractmethod
class Vehicle:
  @abstractmethod
  def start(self):
    pass

  def stop(self):
    pass

class ElectricCar(Vehicle):
  def __init__(self,brand):
    self.brand=brand

  def start(self):
    print(f'{self.brand} electric engine starting silently...')

  def stop(self):
    print(f'{self.brand} has stopped. Battery saved.')

class GasCar(Vehicle):
  def __init__(self,brand):
    self.brand=brand

  def start(self):
    print(f'{self.brand} gas engine roaring to life!')

  def stop(self):
    print(f'{self.brand} engine turned off.')

cars=[ElectricCar("Tesla"),GasCar("Toyota")]
for car in cars:
  car.start()
  car.stop()
````

### B2 — Complete GradeBook System

Build a `GradeBook` class that manages students and their grades. The class should have:

- **Attribute:** `students` — a dictionary `{name: [list of grades]}`

- **Methods:**
  1. `add_student(name)` — adds a student with an empty grade list. If the student already exists, print a warning.
  2. `add_grade(name, grade)` — adds a grade (must be between 0 and 100) to the student. If the student doesn't exist, print an error. If the grade is invalid, print an error.
  3. `get_average(name)` — returns the average grade for a student. Handle the case where the student has no grades (return 0).
  4. `get_top_students(n=3)` — returns a sorted list of the top `n` students by average grade (as a list of tuples `(name, average)`).
  5. `summary()` — prints all students with their grades and average.

**Test with:**
- Add at least **5 students**.
- Add **2–4 grades** to each student.
- Print the summary.
- Print the **top 3** students.

````python
class GradeBook:
  def __init__(self):
    self.students={}

  def add_student(self, name):
    if name in self.students:
      print(f"Warning: Student {name} already exists.")
    else:
      self.students[name]=[]

  def add_grade(self, name, grade):
    if name not in self.students:
      print(f"Error: Student {name} doesn't exist.")
    elif not (0 <= grade <= 100):
      print(f"Error: Grade {grade} is invalid.")
    else:
      self.students[name].append(grade)

  def get_average(self, name):
    if name not in self.students or not self.students[name]:
      return 0
    return sum(self.students[name]) / len(self.students[name])

  def get_top_students(self, n=3):
    sorted_students = sorted([(name, self.get_average(name)) for name in self.students],key=lambda x: x[1],reverse=True)
    return sorted_students[:n]

  def summary(self):
    print("--- GradeBook Summary ---")
    for name, grades in self.students.items():
      avg = self.get_average(name)
      print(f"Student: {name}, Grades: {grades}, Average: {avg:.2f}")

gb = GradeBook()

for name in ["Ali", "Sara", "Omar", "Lama", "Nouf"]:
  gb.add_student(name)

gb.add_grade("Ali", 88)
gb.add_grade("Ali", 92)
gb.add_grade("Sara", 95)
gb.add_grade("Sara", 98)
gb.add_grade("Sara", 100)
gb.add_grade("Omar", 70)
gb.add_grade("Omar", 65)
gb.add_grade("Omar", 80)
gb.add_grade("Lama", 90)
gb.add_grade("Lama", 92)
gb.add_grade("Lama", 88)
gb.add_grade("Nouf", 85)
gb.add_grade("Nouf", 90)

print()
gb.summary()

print()
top_students = gb.get_top_students(3)
print(f"Top 3 Students: {top_students}")
````
