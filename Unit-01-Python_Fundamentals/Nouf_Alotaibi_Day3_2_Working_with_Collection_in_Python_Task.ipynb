## **Lists Challenges & Solutions**

### Quick Review: Lists in Python
- **Ordered**: Items have a defined order.
- **Mutable**: You can change, add, or remove items.
- Can store mixed data types.
- Created using square brackets `[]`.
- Access elements using index, starting at 0.
- Common methods: `.append()`, `.remove()`, `.insert()`, `.sort()`.

**Lists Challenge 3:** Given a list `nums = [5, 10, 15, 20]`, replace the second element with 99.

```python
nums=[5,10,15,20]
nums[1]=99
nums
```

**Lists Challenge 4:** Append the word 'Python' to an empty list and print the list.

**Solution:**

## **Dictionaries Challenges & Solutions**

### Quick Review: Dictionaries in Python
- Store data as **key-value pairs**.
- Keys must be unique and immutable.
- Values can be any data type.
- Created using curly braces `{}` with `key: value` pairs.
- Access values using keys.
- Common methods: `.keys()`, `.values()`, `.items()`.

**Dictionaries Challenge 1:** Create a dictionary for a student with keys: name, age, grade.

**Solution:**

```python
student={
    'name':'Ali',
    'age':22,
    'grade':98
}
student
```

**Dictionaries Challenge 2:** Add a key 'city' with value 'Riyadh' to the dictionary.

**Solution:**

```python
student['city']='Riyadh'
student
```

**Dictionaries Challenge 3:** Remove the key 'grade' from the dictionary.

**Solution:**

```python
student.pop('grade')
student
```

**Dictionaries Challenge 4:** Given `scores = {'Ali': 85, 'Sara': 90}`, print Sara's score.

**Solution:**

```python
scores={
    'Ali':85,
    'Sara':90
}
scores['Sara']
```

**Dictionaries Challenge 5:** Loop through a dictionary and print all keys.

**Solution:**

```python
plants={
    'flowers':'lillies',
    'weeds':'nettles',
    'herbs': 'oregano',
    'trees':'oak'
}
for plant in plants:
  print(plant)
```

## **Sets Challenges & Solutions**

### Quick Review: Sets in Python
- **Unordered**: No defined order.
- **Unique elements**: No duplicates allowed.
- Mutable: Can add or remove elements.
- Created using curly braces `{}` or `set()`.
- No indexing; use loops to access elements.
- Supports mathematical set operations like union and intersection.

**Sets Challenge 1:** Create a set with numbers 1, 2, 3 and print it.

**Dictionaries Challenge (Advanced):** Store employees in a list of dictionaries (name, department, salary) and print names in IT with salary > 8000.

**Solution:**

```python
employees=[
    {'name':'Sara','department':'IT','salary':6500},
    {'name':'Nora','department':'IT','salary':9000},
    {'name':'Ali','department':'Marketing','salary':7000},
    {'name':'Muna','department':'HR','salary':8000},
    {'name':'Ahmed','department':'Manager','salary':10000},
    {'name':'Bader','department':'Marketing','salary':7000},
    {'name':'Reem','department':'Analyst','salary':8500}]
for emp in employees:
  if emp['department']=='IT' and emp['salary']>8000:
    print(emp['name'])
```

**Dictionaries Challenge (Advanced):** Count how many times each word appears in a sentence.

**Solution:**

```python
def count_words(sentence):
    word_counts = {}
    for word in sentence.split():
        if word in word_counts:
            word_counts[word] += 1
        else:
            word_counts[word] = 1
    return word_counts

text = "apple banana apple orange banana apple"
print(count_words(text))
```

**Sets Challenge (Advanced):** Given two sets of students in different subjects, find students in both, in only one, and in either.

**Solution:**

```python
students_in_IT={'Nora','Ali','Sara','Nouf','Reem','Noor'}
students_in_AI={'Sara','Nouf','Bader','Muna','Ahmed','Ali'}
print('students in both: ',students_in_IT.intersection(students_in_AI))
print('students in only one: ',students_in_IT.symmetric_difference(students_in_AI))
print('students in either: ', students_in_IT.union(students_in_AI))
```

**Lists & Tuples  Challenge (Advanced):** Generate even numbers from 50 to 100 and store their squares in a tuple.

**Solution:**

```python
nums=list(range(50,101,2))
squarednums=list(map(lambda x:x**2, nums))
squarednums=tuple(squarednums)
print(squarednums)
```

**Dictionaries & Lists  Challenge (Advanced):** Inventory management: store products in a dictionary with quantities, add a product, update quantity, and print low stock (<5).

**Solution:**

```python
dic={
    'apple':5,
    'kiwi':2,
    'orange':10,
    'banana':3
}

def add_product():
  product=input('add product: ')
  quantity=int(input('add quantity: '))
  dic[product]=quantity
  print(dic)

def update_quantity():
  product = input("Enter product to update: ")
  quantity = int(input("Enter new quantity: "))
  dic[product] = quantity
  print(dic)

def print_low_stock():
  for product,quantity in dic.items():
    if quantity<5:
      print('low stock products: ',product,quantity)

choice=input('what operation would you like to perform? (add,update,print low stock) ')
if choice=='add':
  add_product()
elif choice=='update':
  update_quantity()
elif choice=='print low stock':
  print_low_stock()
else:
  print('please enter a valid choice')
```

**Dictionaries & Lists  Challenge (Advanced):** Store daily sales in a
dictionary (day: list of sales) and calculate total sales for each day.

**Solution:**

```python
sales = {
    "Monday": [100, 200, 150],
    "Tuesday": [80, 120, 90],
    "Wednesday": [300, 100]
}

for day, amounts in sales.items():
    total = sum(amounts)
    print(day, "total Sales:", total)
```
