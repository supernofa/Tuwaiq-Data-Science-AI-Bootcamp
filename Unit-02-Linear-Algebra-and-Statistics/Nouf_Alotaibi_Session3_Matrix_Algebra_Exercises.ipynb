```python
import numpy as np
```

```python
A = np.array([[1, 3],
              [2, 4]])

B = np.array([[2, 0],
              [1, 2]])
```

*   Compute A + B
* Compute A @ B
* Compute B @ A
* Are A @ B and B @ A equal?




```python
print(A+B)
print(A@B)
print(B@A)
# A @ B and B @ A are not equal
```

#Determinant & Inverse

1.   Compute determinant of A
2. Compute inverse of A
3. Verify A @ A⁻¹ = I




```python
det_A=np.linalg.det(A)
A_inv=np.linalg.inv(A)
print(det_A)
print(A_inv)
print(A@A_inv)
```

#Solve a System

Solve:

* 3x + 2y = 16
* 4x − y = 9

Tasks:

1. Write matrix A and vector b

2. Use np.linalg.solve

3. Verify by substitution

```python
A=np.array([[3,2],
            [4,-1]])
b=np.array([16,9])
x=np.linalg.solve(A,b)
print(x)
#verifying by substitution
x,y=np.linalg.solve(A,b)
print(3*x+2*y)
print(4*x-y)
```

#Rank Investigation

```python
M = np.array([[1, 2, 3],
              [2, 4, 6],
              [1, 0, 1]])
```

1. Compute rank

2. Are rows independent?

3. Explain the result

```python
m=np.linalg.matrix_rank(M)
print(m)
# 2 rows are linearly independent
```
