#Compute Eigenvalues

````python
import numpy as np
````

````python
A = np.array([[4, 2],
              [1, 3]])
````

1. Compute eigenvalues

2. Compute eigenvectors

````python
eigenvalues,eigenvectors = np.linalg.eig(A)
print(eigenvalues,eigenvectors)
````

#Verify Eigen Equation

For each eigenpair:

1. Compute A @ v

2. Compute λ * v

3. Compare results

````python
print('A@v=\n',A@eigenvectors)
print('A@v=\n',eigenvalues*eigenvectors)
#both result are the same
````

Manual Characteristic Equation

For matrix:



```
np.array([[3, 1],
          [0, 2]])
```



1- Compute determinant of (A − λI)

2- Solve quadratic equation

````python
import sympy as sp
B = sp.Matrix([[3, 1],
              [0, 2]])

λ = sp.symbols('λ')
char_poly = (B - λ*sp.eye(2)).det()
sp.solve(char_poly, λ)
````

#Diagonalization

1. Compute P, D

2. Reconstruct A

3. Verify accuracy

````python
eigenvalues_A, eigenvectors_A = np.linalg.eig(A)
P=eigenvectors_A
D=np.diag(eigenvalues_A)
P_inv=np.linalg.inv(P)
print("Reconstructed A = \n", P @ D @ P_inv)
````

#PCA Practice

````python
X = np.array([[1,2],
              [2,3],
              [3,4],
              [4,5]])
````

1. Center data

2. Compute covariance matrix

3. Compute eigenvalues

4. Identify principal direction

````python

X = np.array([
    [1, 2],
    [2, 3],
    [3, 4],
    [4, 5]
])

X_centered = X - np.mean(X, axis=0)
print("Centered data:",X_centered)
cov_matrix = np.cov(X_centered, rowvar=False)
print("\nCovariance matrix:",cov_matrix)
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
print("\nEigenvalues:",eigenvalues)
print("\nEigenvectors:",eigenvectors)
principal_index = np.argmax(eigenvalues)
principal_direction = eigenvectors[:, principal_index]
print("\nPrincipal direction:",principal_direction)
````
