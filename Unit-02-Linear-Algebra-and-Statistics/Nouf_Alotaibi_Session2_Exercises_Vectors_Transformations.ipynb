#Exercise 1 — Vector basics & shapes

## Task

Create vectors:

v = [2, -1, 3]

w = [1, 4, -2]

* Print:

* shape, ndim

* Convert v into a column vector shape (3,1)

```python
import numpy as np

v=np.array([2,-1,3])
w=np.array([1,4,-2])
print('shape: ',v.shape,',',w.shape)
print('ndim: ',v.ndim,',',w.ndim)
print('converting to (3,1): \n',v.reshape(3,1))
```

# Exercise 2 — Norm & normalization

## Task

* Compute ||v||

* Create a normalized vector v_unit = v / ||v||



```python
norm_v=np.linalg.norm(v)
print('||v||: ',norm_v)
v_unit=v/norm_v
print(v_unit)
```

# Exercise 3 — Dot product & cosine similarity

## Task

* Compute dot product v·w

* Compute cosine similarity

* Interpret:

Is similarity closer to 1, 0, or -1?

```python
dot=v@w
print('dot product: ',dot)
cos_sim=dot/(np.linalg.norm(v) * np.linalg.norm(w))
print('cosine similraity: ',cos_sim)
```
