#Exercise 5 – Matrix Multiplication

Given:
```
A = np.array([[1, 2],
              [3, 4]])


B = np.array([[2, 0],
              [1, 2]])
```

* Compute element-wise multiplication.
* Compute matrix multiplication.
* Explain the difference in a comment.

````python
A = np.array([[1, 2],
              [3, 4]])


B = np.array([[2, 0],
              [1, 2]])
print(A*B)
print(A@B)
# with element wise multiplication (*) each element in matrix A is multipled by the element matching it's position in matrix B
# with matrix multiplication (@) rows of matrix A are multiplied by columns in matrix B
````

#Exercise 6 – Statistics Practice

Create a dataset of exam scores:

```
scores = np.array([
    [70, 80, 90],
    [85, 75, 95],
    [60, 65, 70],
    [90, 88, 92]
])

```

* Compute mean score per student.

* Compute mean score per subject.

* Find the highest and lowest score.

* Compute standard deviation.

````python
scores = np.array([
    [70, 80, 90],
    [85, 75, 95],
    [60, 65, 70],
    [90, 88, 92]
])
print(scores.mean(axis=1))
print(scores.mean(axis=0))
print(np.max(scores))
print(np.min(scores))
print(np.std(scores))
````
