1- Change the number of filters:

Conv2D(32 → 64 → 128)


Compare accuracy and training time.




```python
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train = x_train / 255.0
x_test  = x_test / 255.0
x_train = x_train.reshape(-1, 28, 28, 1)
x_test  = x_test.reshape(-1, 28, 28, 1)
print(x_train.shape, x_test.shape)
```

```python
model32 = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    tf.keras.layers.MaxPooling2D((2,2)),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

model32.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy'])

history_32 = model32.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.1)

test_loss, test_acc = model32.evaluate(x_test, y_test)
print("test accuracy:", test_acc)
```

```python
model32 = tf.keras.Sequential([
    tf.keras.layers.Conv2D(64, (3,3), activation='relu', input_shape=(28,28,1)),
    tf.keras.layers.MaxPooling2D((2,2)),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

model32.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy'])

history_32 = model32.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.1)

test_loss, test_acc = model32.evaluate(x_test, y_test)
print("test accuracy:", test_acc)
```

```python
model32 = tf.keras.Sequential([
    tf.keras.layers.Conv2D(128, (3,3), activation='relu', input_shape=(28,28,1)),
    tf.keras.layers.MaxPooling2D((2,2)),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

model32.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy'])

history_32 = model32.fit(
    x_train, y_train,
    epochs=5,
    validation_split=0.1)

test_loss, test_acc = model32.evaluate(x_test, y_test)
print("test accuracy:", test_acc)
```

2- Replace MaxPooling with:

AveragePooling2D


Compare results.

```python
model= tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    tf.keras.layers.AveragePooling2D(2, 2),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy'])

model.summary()
```

```python
history= model.fit(
    x_train, y_train,
    epochs=5,
    validation_split=0.1)
```

3- Add Dropout:

tf.keras.layers.Dropout(0.5)


Observe overfitting reduction.

```python
model= tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    tf.keras.layers.MaxPooling2D(2,2),
    tf.keras.layers.Dropout(0.5),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy'])

model.summary()
```

```python
history= model.fit(
    x_train, y_train,
    epochs=5,
    validation_split=0.1)
```
